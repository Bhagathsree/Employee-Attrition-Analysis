<h2 style="color:#1F4E79;">🔄 Bivariate Analysis</h2>

Bivariate Analysis examines the relationship between two variables to identify patterns, associations, and potential factors influencing employee attrition. In this project, attrition is analyzed against selected demographic and job-related variables to understand how different factors may impact employee turnover.

<h3 style="color:#2E75B6;">📈 Attrition vs OverTime</h3>

```python
attrition_overtime = pd.crosstab(
    df["OverTime"],
    df["Attrition"],
    margins=True
)

attrition_overtime
```

```python
attrition_overtime_percent = pd.crosstab(
    df["OverTime"],
    df["Attrition"],
    normalize="index"
).round(3) * 100

attrition_overtime_percent
```

### Observation

The relationship between **OverTime** and **Attrition** reveals the following:

- Among employees **not working overtime**, **89.6%** remained with the organization, while only **10.4%** left.
- Among employees **working overtime**, **69.5%** remained, whereas **30.5%** left the organization.
- The attrition rate of employees working overtime (**30.5%**) is nearly three times higher than that of employees not working overtime (**10.4%**).
- This indicates a clear association between overtime work and employee attrition.

The analysis suggests that employees who work overtime are considerably more likely to leave the organization than those who do not work overtime.

<h3 style="color:#2E75B6;">🧩 Multivariate Analysis</h3>

Multivariate Analysis examines the relationship among three or more variables simultaneously to identify complex patterns that cannot be observed through univariate or bivariate analysis alone. In this project, multivariate analysis is used to understand how multiple employee-related factors collectively influence attrition and workforce characteristics.

### Analysis 1: Correlation Matrix

```python
corr = df.select_dtypes(include=['int64', 'float64']).corr().round(2)

corr
```

### Observation

The correlation matrix reveals the strength and direction of relationships among the numerical variables in the dataset. The key observations are:

- **Job Level** and **Monthly Income** have a **very strong positive correlation (0.95)**, indicating that employees at higher job levels generally earn higher salaries.
- **Job Level** and **Total Working Years** show a **strong positive correlation (0.78)**, suggesting that employees with more work experience tend to hold higher job positions.
- **Total Working Years** is positively correlated with **Age (0.68)**, indicating that older employees generally have greater work experience.
- **Years at Company** has a strong positive relationship with **Years with Current Manager (0.77)** and **Years in Current Role (0.76)**, showing that employees who remain longer in the organization typically spend more time in the same role and under the same manager.
- Most other numerical variables have weak correlations, indicating limited linear relationships among them.

### Analysis 2: Department × OverTime → Average Monthly Income (Pivot Table)

```python
pivot_income = pd.pivot_table(
    df,
    values="MonthlyIncome",
    index="Department",
    columns="OverTime",
    aggfunc="mean"
).round(2)

pivot_income
```

### Observation

The pivot table compares the average monthly income of employees across different departments based on their overtime status.

- Employees in the **Sales** department have the highest average monthly income, regardless of whether they work overtime.
- In the **Human Resources** department, employees who work overtime earn a noticeably higher average monthly income than those who do not.
- In the **Research & Development** department, the average monthly income remains nearly the same for employees with and without overtime, indicating minimal income variation based on overtime status.
- Overall, overtime is associated with a slight increase in average monthly income in some departments, but the effect is not consistent across all departments.
